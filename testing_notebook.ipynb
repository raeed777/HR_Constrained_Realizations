```python
import importlib, helper_tools, Data, Observed_Data, Reconstructed_Field, psi_solver, pspectra_camb, Box, Cosmology, Obs_Geometry
importlib.reload(helper_tools)
importlib.reload(Data)
importlib.reload(Observed_Data)
importlib.reload(Reconstructed_Field)
importlib.reload(psi_solver)
importlib.reload(pspectra_camb)
importlib.reload(Box)
importlib.reload(Cosmology)
importlib.reload(Obs_Geometry)

from Data import Data
from Reconstructed_Field import ReconstructedField
from psi_solver import Operators, make_precond_Sphi_spectral, Wiener_solve_psi_radial_fft, Constrained_realization_psi_radial_fft
import pickle, time
from helper_tools import make_cone_mask_with_subcones, plot_mask_3d, plot_mask_slices, subcone_footprints_on_slice
from pspectra_camb import build_camb_pk_callable
from Box import Box
from Cosmology import Cosmology
import numpy as np
from Observed_Data import ObservedData
from Obs_Geometry import Obs_Geometry
from psi_solver import _get_sigma_x, _make_Wx, make_matvec_and_rhs_psi_radial_fft, pcg
import random as rng
box = Box(64, 250)

from Cosmology import make_planck18_cosmology
z0 = 0.0
cosmology = make_planck18_cosmology(z0)
# pick a safe kmax in h/Mpc for your box (≈ 1.1 * sqrt(3) * pi/dx)
kmax_h = 1.1 * np.sqrt(3.0) * np.pi / box.dx
print('max k: ', kmax_h)

# Build CAMB ONCE and get the callable
Pk_callable, _ = build_camb_pk_callable(
    Om=cosmology.Om, Ob=cosmology.Ob, h=cosmology.h, ns=cosmology.ns,
    sigma8_target=cosmology.sigma8,
    z=0.0,
    kmax_h=kmax_h,
    nonlinear=False
)

M, dbg = make_cone_mask_with_subcones(
    box,
    los_dir="z",
    cone_half_angle_deg=11.0,
    observer_offset_L=1.5,
    subcone_count=10,
    subcone_half_angle_deg=1.7,
    subcone_layout="random",   # <—
    seed=42,
    return_debug=True
)
print("coverage:", M.mean())

# # Compute footprints for central z-slice
# foots = subcone_footprints_on_slice(dbg, box, slice_axis='z', slice_index=n//2,
#                                     use_ellipse=False)  # change to True for ellipses

M = None
radial_observer_offset_L = 1.0

obs_geo = Obs_Geometry(box)

obs_geo.set_observer(radial_observer_offset_L)
obs_geo.initialize()



def debug_sigma_scaling(
    ops,
    obs_data_base,
    sigma_fracs,
    include_geom=True,
    rtol=1e-6,
    maxit=300,
):
    """
    For a *fixed* underlying truth (obs_data_base.delta_r, etc.),
    vary the noise level sigma_frac and measure how the Wiener
    solution norm and rhs norm scale.

    This uses the full A_psi and rhs, not a toy system.
    """
    import copy

    results = []

    true_delta = obs_data_base.delta_r.copy()
    print("||delta_true|| =", np.linalg.norm(true_delta))

    for sf in sigma_fracs:
        print("\n=== sigma_frac =", sf, "===")

        # Clone ObservedData so we don't overwrite the base
        obs = copy.deepcopy(obs_data_base)

        # Regenerate rsd field if you want the same signal each time:
        # (comment out if you want to keep the same delta_s_r)
        # obs.generate_mock_fields(obs_geo, rng=np.random.default_rng(), Pk_callable=Pk_callable)

        # New noise with this sigma_frac
        obs.generate_artificial_noise(sigma_frac=sf, jitter_frac=0.0)
        obs.add_noise_and_mask()

        sigma_x = _get_sigma_x(obs)
        W_x     = _make_Wx(sigma_x, M=getattr(obs, "mask", None), eps=1e-30)

        print("  <sigma_x^2> =", np.mean(sigma_x**2))
        print("  <W_x>       =", np.mean(W_x))
        print("  ||d||       =", np.linalg.norm(obs.d))

        # Build A and rhs for this data
        A, rhs = make_matvec_and_rhs_psi_radial_fft(
            ops,
            b_bias=float(obs.b_bias),
            sigma_x=sigma_x,
            d=obs.d,
            M=obs.mask,
            eps=1e-30,
            include_geom=include_geom,
        )

        print("  ||rhs||     =", np.linalg.norm(rhs))

        # Turn OFF preconditioner for this diagnostic to avoid confusion
        psi = pcg(A, rhs, apply_Minv=None, rtol=rtol, maxit=maxit, verbose=False)
        phi = ops.apply_Sphi_sqrt_fft(psi)

        print("  ||psi||     =", np.linalg.norm(psi))
        print("  ||phi||     =", np.linalg.norm(phi))

        ratio = np.linalg.norm(phi) / (np.linalg.norm(true_delta) + 1e-30)
        print("  ||phi|| / ||delta_true|| =", ratio)

        results.append((sf, ratio))

    return results

# sigma_fracs = [1.5, 3.0, 6.0, 12.0, 24.0]

# obs_data = ObservedData(box, cosmology, mask=M, b_bias=1.7, dataset="radial")
# obs_data.generate_mock_fields(obs_geo, rng=np.random.default_rng(), Pk_callable=Pk_callable)

# ops = Operators(box, cosmology, Pdelta_callable=Pk_callable, obs_geometry=obs_geo)

# results = debug_sigma_scaling(ops, obs_data, sigma_fracs)
# print("summary:", results)

x = np.random.normal(size=phi.shape)
y = np.random.normal(size=phi.shape)

Ly  = ops.apply_L_rsd_radial_fft(y, include_geom=True)
LTx = ops.apply_Lt_rsd_radial_fft(x, include_geom=True)

lhs = np.vdot(x, Ly).real
rhs = np.vdot(LTx, y).real
print(lhs, rhs, (lhs-rhs)/lhs)

```
